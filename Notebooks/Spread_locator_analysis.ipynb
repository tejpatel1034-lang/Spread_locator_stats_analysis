### 1. What are Statistical Distributions?

**Definition:** A statistical distribution describes how the values of a variable are spread or distributed across different possible outcomes and how frequently or likely those values occur.

**Explanation:** Statistical distributions help us understand the pattern of data, including which values occur more frequently, how much the data varies, and whether the data is concentrated around a particular value or spread over a wide range.

Statistical distributions are mainly classified into two types:

1. **Discrete Distribution:** Used for countable values. Examples include Bernoulli, Binomial, and Poisson distributions.
2. **Continuous Distribution:** Used for measurable values within a range. Examples include Normal, Log-Normal, and Power Law distributions.

**Example:** In the *Spread Locator: A Statistical Distribution Analysis Model* project, statistical distributions are used to understand transaction amounts, identify patterns in financial data, and detect unusually large transactions.

**Conclusion:** Statistical distributions are important tools for data analysis, probability estimation, decision-making, and understanding the behaviour of data.

### 2. What is a Q-Q Plot and Why Is It Used?

**Definition:** A Q-Q (Quantile-Quantile) Plot is a graphical tool used to compare the distribution of a dataset with a theoretical distribution or another dataset.

**Explanation:** A Q-Q plot compares the quantiles of observed data with the quantiles expected from a theoretical distribution, such as the Normal Distribution. It helps us understand whether the data follows the expected distribution.

**Why is a Q-Q Plot Used?**

1. **Check Normality:** To determine whether data follows a Normal Distribution.
2. **Identify Outliers:** To identify unusual values that deviate from the expected pattern.
3. **Compare Distributions:** To compare the shape of observed data with a theoretical distribution.
4. **Assess Model Fit:** To evaluate how well a statistical distribution represents the data.

**Interpretation:**
- If the points lie approximately along a straight line, the distributions are similar.
- If the points deviate significantly from the line, the data may follow a different distribution.

**Example:** In the *Spread Locator: A Statistical Distribution Analysis Model* project, a Q-Q plot is used to check whether transaction amounts follow a Normal Distribution and whether their logarithms are approximately normally distributed.

**Conclusion:** A Q-Q plot is a useful visual tool for checking distributional assumptions and evaluating the suitability of statistical models.

### 4. What Is Bernoulli Distribution?

**Definition:** Bernoulli Distribution is a discrete probability distribution that represents a random experiment with only two possible outcomes: success (1) or failure (0).

**Explanation:** The Bernoulli Distribution is used when an experiment has exactly two possible outcomes. The probability of success is denoted by \(p\), and the probability of failure is \(1-p\).

The probability mass function (PMF) is:

\[
P(X=x)=p^x(1-p)^{1-x}
\]

Where:
- \(X\) = Random variable (0 or 1)
- \(p\) = Probability of success
- \(1-p\) = Probability of failure

**Mean and Variance:**

- Mean: \(E(X)=p\)
- Variance: \(\operatorname{Var}(X)=p(1-p)\)

**Example:** Suppose a transaction is classified as either suspicious or not suspicious. If the probability of a transaction being suspicious is 0.2, then:

- Success (1): Suspicious transaction, probability = 0.2.
- Failure (0): Non-suspicious transaction, probability = 0.8.

**Applications:**
1. Quality control and defect detection.
2. Fraud detection in financial transactions.
3. Yes-or-no experiments and binary classification.

**Conclusion:** Bernoulli Distribution is useful for modelling events that have only two possible outcomes, such as success or failure.

### 5. What Is Binomial Distribution?

**Definition:** Binomial Distribution is a discrete probability distribution that calculates the probability of obtaining a specific number of successes in a fixed number of independent trials, where each trial has only two possible outcomes: success or failure.

**Explanation:** The Binomial Distribution is used when an experiment is repeated a fixed number of times, the trials are independent, and the probability of success remains constant.

The probability mass function (PMF) is:

\[
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
\]

Where:
- \(n\) = Total number of trials
- \(k\) = Number of successes
- \(p\) = Probability of success
- \(1-p\) = Probability of failure
- \(\binom{n}{k}\) = Number of ways to obtain \(k\) successes in \(n\) trials

**Mean and Variance:**

- Mean: \(E(X)=np\)
- Variance: \(\operatorname{Var}(X)=np(1-p)\)

**Example:** Suppose 10 transactions are independently checked, and each transaction has a 20% probability of being classified as suspicious. The Binomial Distribution can calculate the probability that exactly 3 out of the 10 transactions are suspicious.

Here, \(n=10\), \(k=3\), and \(p=0.2\).

**Applications:**
1. Fraud detection in financial transactions.
2. Quality control and defect analysis.
3. Calculating success or failure counts.
4. Analysing outcomes of repeated independent experiments.

**Conclusion:** Binomial Distribution is useful for calculating the probability of a specific number of successes in a fixed number of independent trials.

### 6. Explain Log-Normal Distribution

**Definition:** A Log-Normal Distribution is a continuous probability distribution in which the logarithm of a random variable follows a Normal Distribution. It is commonly used for data that contains positive values and is skewed to the right.

**Explanation:** In a Log-Normal Distribution, the original data is not normally distributed, but its logarithmic values are approximately normally distributed. It is useful when most observations have smaller values and a few observations have very large values.

**Mathematical Representation:**

If \(X\) follows a Log-Normal Distribution, then:

\[
\ln(X)\sim N(\mu,\sigma^2)
\]

Where:
- \(X\) = Positive random variable
- \(\ln(X)\) = Natural logarithm of \(X\)
- \(\mu\) = Mean of the logarithmic values
- \(\sigma\) = Standard deviation of the logarithmic values

**Characteristics:**
1. It takes only positive values.
2. Its distribution is generally right-skewed.
3. It is suitable for data with a long right tail.
4. Taking the logarithm of the data can make its distribution closer to Normal.

**Example:** In the *Spread Locator: A Statistical Distribution Analysis Model* project, transaction amounts are analysed using a Log-Normal Distribution because most transactions may be relatively small, while a few transactions can have much larger amounts.

**Applications:**
1. Financial transaction analysis.
2. Income and wealth modelling.
3. Risk assessment.
4. Modelling prices and other positive-valued data.

**Conclusion:** The Log-Normal Distribution is useful for analysing positive, right-skewed data and estimating probabilities associated with unusually large values.

### 7. Explain Power Law Distribution

**Definition:** A Power Law Distribution is a probability distribution in which the frequency or probability of an event decreases as the value of the event increases, following a power function. It is often used to describe data where a small number of observations have extremely large values.

**Explanation:** In a Power Law Distribution, most observations tend to have smaller values, while a few observations have very large values. This creates a long tail in the distribution. Such data is commonly found in financial transactions, wealth distribution, website traffic, and network analysis.

**Mathematical Representation:**

The probability density function (PDF) for a continuous Power Law Distribution can be written as:

\[
f(x)=Cx^{-\alpha},\quad x\geq x_{\min}
\]

Where:
- \(x\) = Observed value
- \(C\) = Normalising constant
- \(\alpha\) = Power law exponent
- \(x_{\min}\) = Minimum value from which the power law applies

**Characteristics:**
1. It is commonly associated with a heavy-tailed distribution.
2. Very large values are rare but can occur.
3. It can describe data with substantial variation in magnitude.
4. Its tail often decreases more slowly than the tail of an exponential distribution.

**Example:** In the *Spread Locator: A Statistical Distribution Analysis Model* project, a Power Law model can be investigated to determine whether a small number of transactions are much larger than most other transactions.

**Applications:**
1. Wealth and income distribution analysis.
2. Financial risk analysis.
3. Network and website traffic analysis.
4. Studying the frequency of large and small events.

**Conclusion:** Power Law Distribution is useful for studying datasets with heavy tails and large differences between typical and extreme values. However, a heavy-tailed appearance alone does not prove that data follows a Power Law Distribution; statistical testing is required.

### 8. What Is Box-Cox Transform?

**Definition:** The Box-Cox Transformation is a statistical technique used to transform positive numerical data to make it more normally distributed, reduce skewness, and improve the performance of certain statistical models.

**Explanation:** Sometimes, data contains extreme values or is highly skewed. The Box-Cox Transformation applies a mathematical power transformation to reduce skewness and make the data more suitable for statistical analysis. It uses a parameter called lambda (\(\lambda\)) to determine the appropriate transformation.

**Mathematical Formula:**

\[
Y(\lambda)=
\begin{cases}
\dfrac{Y^\lambda-1}{\lambda}, & \lambda\neq 0\\[6pt]
\ln(Y), & \lambda=0
\end{cases}
\]

Where:
- \(Y\) = Original positive data value
- \(Y(\lambda)\) = Transformed value
- \(\lambda\) = Transformation parameter

**Characteristics:**
1. It is generally applied to strictly positive data.
2. It helps reduce skewness in a dataset.
3. It can make data closer to a Normal Distribution.
4. The appropriate lambda value is usually estimated from the data.
5. It does not guarantee that the transformed data will be normally distributed.

**Example:** In the *Spread Locator: A Statistical Distribution Analysis Model* project, the Box-Cox Transformation is applied to transaction amounts to reduce right skewness. In our analysis, the original skewness was approximately **3.73**, while the transformed skewness was approximately **−0.01**, indicating a substantial reduction in skewness.

**Applications:**
1. Preparing data for statistical modelling.
2. Reducing skewness in numerical data.
3. Improving the suitability of data for methods that assume normality.
4. Analysing financial and business data.

**Conclusion:** The Box-Cox Transformation is a useful statistical technique for reducing skewness and making positive numerical data more suitable for statistical analysis.

### 9. Explain Poisson Distribution with an Example

**Definition:** Poisson Distribution is a discrete probability distribution used to calculate the probability of a specific number of events occurring within a fixed interval of time, space, or another defined measurement, when events occur independently at a constant average rate.

**Explanation:** The Poisson Distribution is useful for counting how many times an event occurs within a specified interval. It uses the average number of events, represented by lambda (\(\lambda\)).

**Mathematical Formula:**

\[
P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}
\]

Where:
- \(X\) = Number of events
- \(k\) = Specific number of events
- \(\lambda\) = Average number of events per interval
- \(e\) = Mathematical constant approximately equal to 2.718
- \(k!\) = Factorial of \(k\)

**Example:** Suppose a bank receives an average of 3 suspicious transaction alerts per hour. We want to calculate the probability of receiving exactly 2 alerts in one hour.

Here, \(\lambda=3\) and \(k=2\).

\[
P(X=2)=\frac{e^{-3}\times3^2}{2!}
\]

\[
P(X=2)\approx0.2240
\]

Therefore, the probability of receiving exactly 2 alerts in one hour is approximately **22.40%**.

**Characteristics:**
1. It is a discrete probability distribution.
2. It is used to model event counts.
3. Its mean and variance are both equal to \(\lambda\).
4. It assumes a constant average event rate and independent occurrences.

**Applications:**
1. Counting suspicious financial transactions.
2. Analysing customer arrivals at a bank.
3. Counting website visits within a fixed period.
4. Studying equipment failures over time.

**Conclusion:** Poisson Distribution is useful for estimating the probability of a specific number of events occurring within a fixed interval, given their average occurrence rate.

### 10. What Is Z-Score Probability?

**Definition:** A Z-score measures how many standard deviations a data value is above or below the mean of a dataset. Z-score probability uses this score to estimate the probability of observing a value within a particular range, usually assuming the data follows a Normal Distribution.

**Explanation:** The Z-score helps us understand how unusual a value is compared with the average. A positive Z-score indicates a value above the mean, while a negative Z-score indicates a value below the mean.

**Mathematical Formula:**

\[
Z=\frac{X-\mu}{\sigma}
\]

Where:
- \(Z\) = Z-score
- \(X\) = Observed data value
- \(\mu\) = Mean of the distribution
- \(\sigma\) = Standard deviation

**Example:** Suppose the average transaction amount is ₹2,000, the standard deviation is ₹1,000, and a transaction amount is ₹5,000.

\[
Z=\frac{5000-2000}{1000}=3
\]

The transaction has a Z-score of **3**, meaning its amount is three standard deviations above the mean.

If transaction amounts follow a Normal Distribution, approximately **0.135%** of observations lie above a Z-score of 3. This probability estimate depends on the Normal Distribution assumption.

**Applications:**
1. Identifying unusually high or low values.
2. Analysing financial transactions.
3. Detecting potential statistical outliers.
4. Estimating probabilities under a Normal Distribution assumption.

**Conclusion:** Z-score probability helps measure how unusual a value is and estimate its probability relative to a distribution. However, a high Z-score alone does not prove that a transaction is fraudulent.

### 11. Differentiate Probability Density Function (PDF) and Cumulative Distribution Function (CDF)

**Definition:**

- **Probability Density Function (PDF):** A function that describes the density of probability at different values of a continuous random variable.
- **Cumulative Distribution Function (CDF):** A function that gives the probability that a random variable is less than or equal to a particular value.

### Difference Between PDF and CDF

| Basis | PDF | CDF |
|---|---|---|
| Full Form | Probability Density Function | Cumulative Distribution Function |
| Purpose | Describes the density of probability across values. | Calculates cumulative probability up to a value. |
| Formula | \(f(x)\) | \(F(x)=P(X\leq x)\) |
| Probability | Probability over an interval is the area under the PDF curve. | Directly gives the probability up to a specified value. |
| Values | Density cannot be negative. | Values range from 0 to 1. |
| Behaviour | The density can increase or decrease. | The CDF is always non-decreasing. |
| Graph | Shows the shape of the probability distribution. | Shows how probability accumulates. |
| Application | Understanding where values are concentrated. | Calculating probabilities below or above a threshold. |

**Example:** In the *Spread Locator: A Statistical Distribution Analysis Model* project, suppose \(X\) represents a transaction amount.

- **PDF:** Helps us understand the density of transaction amounts around ₹5,000.
- **CDF:** If \(F(5000)=0.86\), then approximately 86% of transactions are expected to be ₹5,000 or less under the fitted model.

The probability of a transaction exceeding ₹5,000 is:

\[
P(X>5000)=1-F(5000)
\]

\[
=1-0.86=0.14=14\%
\]

**Conclusion:** PDF describes probability density, whereas CDF gives cumulative probability. Both functions are important for understanding distributions and estimating probabilities in statistical analysis.

#### Bernoulli Distribution — Transaction Occurrence

### Objective
Model transaction success or failure using a Bernoulli distribution.

### Definition
A Bernoulli random variable has two possible outcomes:
- 1 = Success
- 0 = Failure

The parameter p represents the probability of success.

We estimate p from the observed proportion of successful transactions in the dataset.


```python

# Inspect the actual transaction status categories
print(df["transaction_status"].value_counts(dropna=False))
```

```python

# Standardize transaction status
status = df["transaction_status"].astype("string").str.strip().str.lower()
```

```python
# Convert success/failure into binary values
status_mapping = {
    "success": 1,
    "failed": 0
}

bernoulli_data = status.map(status_mapping)
```

```python
# Check whether any status was not mapped
print("Unmapped status values:")
print(df.loc[bernoulli_data.isna(), "transaction_status"].value_counts(dropna=False))
```

```python
# Keep only recognized success/failure records for this model
bernoulli_data = bernoulli_data.dropna().astype(int)

print("\nTotal valid binary observations:", len(bernoulli_data))
print("Successes:", (bernoulli_data == 1).sum())
print("Failures:", (bernoulli_data == 0).sum())
```

```python

# Estimate probability of success
p_success = bernoulli_data.mean()
p_failure = 1 - p_success

print(f"Estimated probability of success (p): {p_success:.4f}")
print(f"Estimated probability of failure (1-p): {p_failure:.4f}")

print(f"\nSuccess rate: {p_success * 100:.2f}%")
print(f"Failure rate: {p_failure * 100:.2f}%")
```


#### Bernoulli Distribution — Interpretation

The Bernoulli model represents transaction outcomes as binary values.

The estimated success probability is calculated as the number of successful transactions divided by the number of valid success/failure observations.

The estimated failure probability is equal to 1 minus the success probability.



#### Binomial Distribution — Weekly Transaction Count

##### Objective
Compare observed weekly transaction counts with a fitted Binomial distribution.

##### Assumptions
1. The number of trials n is fixed.
2. Each trial has two possible outcomes.
3. The probability of success p is constant.
4. Trials are independent.

The Binomial model is appropriate only if the transaction count represents successes out of a known number of possible trials.


```python

# Inspect transaction_count before fitting a distribution
print(df["transaction_count"].describe())

print("\nObserved transaction count frequencies:")
print(df["transaction_count"].value_counts().sort_index())

print("\nMissing counts:", df["transaction_count"].isna().sum())
print("Negative counts:", (df["transaction_count"] < 0).sum())
```

```python

# Inspect the maximum observed count
max_observed_count = df["transaction_count"].max()

print("Maximum observed transaction count:", max_observed_count)

# Set n_trials ONLY if this is a justified maximum possible count
# Replace this value with the actual number of possible weekly trials
# if your assignment or business definition specifies it.
n_trials = int(max_observed_count)

print("Provisional n_trials:", n_trials)
```

```python
from scipy.stats import binom

# Keep valid integer counts between 0 and n_trials
weekly_counts = pd.to_numeric(
    df["transaction_count"], errors="coerce"
).dropna()

weekly_counts = weekly_counts[
    (weekly_counts >= 0) &
    (weekly_counts <= n_trials) &
    (weekly_counts % 1 == 0)
].astype(int)

if len(weekly_counts) == 0:
    raise ValueError("No valid counts available for the selected n_trials.")
```

```python
# Method-of-moments estimate of p
p_binomial = weekly_counts.mean() / n_trials
p_binomial = float(np.clip(p_binomial, 0, 1))

print("Observations used:", len(weekly_counts))
print("Number of trials (n):", n_trials)
print(f"Estimated probability (p): {p_binomial:.4f}")
print(f"Expected count (n × p): {n_trials * p_binomial:.4f}")
```

```python
#Compare observed and expected frequencies

x = np.arange(n_trials + 1)

observed_freq = np.array([
    (weekly_counts == k).sum()
    for k in x
])

expected_freq = binom.pmf(
    x, n=n_trials, p=p_binomial
) * len(weekly_counts)

binomial_results = pd.DataFrame({
    "Weekly Count": x,
    "Observed Frequency": observed_freq,
    "Expected Frequency": expected_freq.round(3)
})

display(binomial_results)
```

```python

# Create x-axis values matching the frequency data
x = np.arange(len(observed_freq))

plt.figure(figsize=(10, 5))

bars1 = plt.bar(
    x - 0.2,
    observed_freq,
    width=0.4,
    label="Observed"
)

bars2 = plt.bar(
    x + 0.2,
    expected_freq,
    width=0.4,
    label="Binomial Expected"
)

# Add data labels
plt.bar_label(bars1, fmt="%.2f", padding=3)
plt.bar_label(bars2, fmt="%.2f", padding=3)

plt.xlabel("Weekly Transaction Count")
plt.ylabel("Frequency")
plt.title("Observed vs Fitted Binomial Distribution")

plt.xticks(x)
plt.legend()
plt.tight_layout()
plt.show()

```


### Binomial Distribution — Interpretation

The observed frequencies show how often each weekly transaction count occurs in the dataset.

The expected frequencies are calculated from the fitted Binomial model using the selected number of trials n and estimated probability p.

Differences between observed and expected frequencies indicate how closely the model describes the data. The conclusion depends on whether the Binomial assumptions are appropriate for the underlying transaction process.


### 2. Fit Data to a Poisson Distribution

#### Objective
Fit a Poisson distribution to the number of transactions per day.

#### Methodology
1. Convert transaction dates into datetime format.
2. Count transactions for each day.
3. Include days with zero transactions within the observed date range.
4. Estimate the Poisson parameter lambda (λ).
5. Compare observed and expected frequencies.
6. Visualize the fitted Poisson distribution.

##### Prepare transaction dates

```python

# Check the date column
print(df["transaction_date"].head())

# Convert dates to datetime format
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"],
    errors="coerce"
)

# Count invalid or missing dates
print(
    "Missing or invalid dates:",
    df["transaction_date"].isna().sum()
)
```

##### Calculate daily transaction counts

```python

# Remove records without valid transaction dates
valid_dates = df.dropna(subset=["transaction_date"]).copy()

# Count transactions for each calendar day
daily_counts = (
    valid_dates
    .set_index("transaction_date")
    .resample("D")
    .size()
)

print("Daily transaction counts:")
display(daily_counts.head(10))

print("\nTotal days:", len(daily_counts))
print("Days with zero transactions:", (daily_counts == 0).sum())
```

```python
from scipy.stats import poisson

lambda_hat = daily_counts.mean()

print(f"Estimated lambda (λ): {lambda_hat:.4f}")
print(f"Average transactions per day: {lambda_hat:.2f}")

print(f"Daily count variance: {daily_counts.var():.4f}")
print(f"Daily count mean: {daily_counts.mean():.4f}")
```

```python

# Possible daily counts: 0 up to the maximum observed count
k_values = np.arange(int(daily_counts.max()) + 1)
```

```python
# Observed frequency of each daily count
observed_freq = np.array([
    (daily_counts == k).sum()
    for k in k_values
])
```

```python
# Expected frequency under the fitted Poisson model
expected_freq = (
    poisson.pmf(k_values, mu=lambda_hat)
    * len(daily_counts)
)

poisson_results = pd.DataFrame({
    "Transactions per Day": k_values,
    "Observed Days": observed_freq,
    "Expected Days (Poisson)": expected_freq.round(3)
})

display(poisson_results)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

bars1 = plt.bar(
    k_values - 0.2,
    observed_freq,
    width=0.4,
    label="Observed"
)

bars2 = plt.bar(
    k_values + 0.2,
    expected_freq,
    width=0.4,
    label="Fitted Poisson"
)

# Add data labels above each bar
plt.bar_label(bars1, fmt="%.2f", padding=3)
plt.bar_label(bars2, fmt="%.2f", padding=3)

plt.xlabel("Number of Transactions per Day")
plt.ylabel("Number of Days")
plt.title("Observed vs Fitted Poisson Distribution")

plt.xticks(k_values)
plt.legend()
plt.tight_layout()
plt.show()


```


##### Poisson Distribution — Conclusion

The Poisson distribution was fitted to daily transaction counts by estimating λ as the sample mean of the daily counts.

The observed daily frequencies were compared with the expected frequencies generated by the fitted Poisson model.

The suitability of the model should be evaluated using the observed-versus-expected plot, the relationship between the sample mean and variance, and the assumption that events occur at a reasonably stable rate over the observation period.

### 4. Q-Q Plot: Normality Testing

A Q-Q plot compares the quantiles of observed data with the quantiles of a theoretical normal distribution.

If the points approximately follow the reference line, the data may be normally distributed. Systematic deviations from the line indicate non-normality.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Use the positive transaction amounts prepared earlier
amounts_clean = amounts.dropna()
amounts_clean = amounts_clean[amounts_clean > 0]

# Log-transform the amounts
log_amounts = np.log(amounts_clean)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Q-Q plot for raw amounts
stats.probplot(amounts_clean, dist="norm", plot=axes[0])
axes[0].set_title("Q-Q Plot: Raw Transaction Amounts")

# Q-Q plot for log-transformed amounts
stats.probplot(log_amounts, dist="norm", plot=axes[1])
axes[1].set_title("Q-Q Plot: Log-Transformed Amounts")

plt.tight_layout()
plt.show()
```

```python
sample = amounts_clean.sample(
    n=min(5000, len(amounts_clean)),
    random_state=42
)

stat_raw, p_raw = stats.shapiro(sample)

log_sample = np.log(sample)
stat_log, p_log = stats.shapiro(log_sample)

print("Raw amounts - Shapiro-Wilk p-value:", p_raw)
print("Log amounts - Shapiro-Wilk p-value:", p_log)

print("\nAt significance level 0.05:")
print("Raw amounts:", "Reject normality" if p_raw < 0.05 else "Do not reject normality")
print("Log amounts:", "Reject normality" if p_log < 0.05 else "Do not reject normality")
```

##### Conclusion: Q-Q Plot and Normality Testing

The Shapiro-Wilk normality test was performed on the original and log-transformed transaction amounts using a significance level of 0.05.

The raw transaction amounts produced a p-value of (1.91 \times 10^{-18}), which is below 0.05. Therefore, the null hypothesis of normality was rejected.

After applying the natural logarithm transformation, the p-value increased to 0.1061, which is above 0.05. Therefore, the null hypothesis of normality was not rejected for the log-transformed amounts.

These findings suggest that the original transaction amounts are not normally distributed, while their logarithms are more consistent with a normal distribution. This supports the Log-Normal model as a reasonable candidate for modeling transaction amounts, although additional goodness-of-fit checks may be required.

### 5. Box-Cox Transformation

**Objective:** Apply the Box-Cox transformation to positive transaction amounts to reduce skewness and potentially stabilize variance.

The optimal lambda parameter is estimated from the data using maximum likelihood.

```python
from scipy.stats import boxcox, skew

# Use valid positive transaction amounts
amounts_clean = amounts.dropna()
amounts_clean = amounts_clean[amounts_clean > 0]

# Apply Box-Cox transformation
boxcox_data, lambda_value = boxcox(amounts_clean)

print("Optimal Box-Cox lambda:", lambda_value)
print("Original variance:", np.var(amounts_clean, ddof=1))
print("Transformed variance:", np.var(boxcox_data, ddof=1))

print("\nOriginal skewness:", skew(amounts_clean))
print("Transformed skewness:", skew(boxcox_data))
```

##### Conclusion: Box-Cox Transformation

The Box-Cox transformation was applied to the positive transaction amounts to reduce skewness and improve distributional symmetry.

The estimated optimal lambda was -0.1808. Before transformation, the transaction amounts had a skewness of 3.7301, indicating strong positive skewness. After transformation, skewness decreased to -0.0106, indicating that the transformed data are approximately symmetric.

The variance decreased from 3,943,025.9720 to 0.01238. However, because the transformation changes the measurement scale, this variance difference alone does not establish variance stabilization.

Overall, the Box-Cox transformation substantially reduced skewness and improved the symmetry of the transaction amount distribution.

### 7. PDF and CDF Analysis of Transaction Amounts

**Objective:** Visualize the Probability Density Function (PDF) and Cumulative Distribution Function (CDF) of transaction amounts using the fitted Log-Normal distribution.

- PDF describes the relative density of transaction amounts across different values.
- CDF gives the probability that a transaction amount is less than or equal to a specified threshold.

```python

# Clean positive transaction amounts
amounts_clean = amounts.dropna()
amounts_clean = amounts_clean[amounts_clean > 0]

# Fit Log-Normal distribution
shape_ln, loc_ln, scale_ln = lognorm.fit(amounts_clean, floc=0)

# Values for plotting
x = np.linspace(amounts_clean.min(), amounts_clean.quantile(0.99), 500)

# Calculate PDF and CDF
pdf_values = lognorm.pdf(
    x, shape_ln, loc=loc_ln, scale=scale_ln
)

cdf_values = lognorm.cdf(
    x, shape_ln, loc=loc_ln, scale=scale_ln
)

# Create two plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# PDF plot: histogram + fitted density
axes[0].hist(
    amounts_clean,
    bins=30,
    density=True,
    alpha=0.5,
    edgecolor="black",
    label="Observed amounts"
)
axes[0].plot(
    x, pdf_values,
    linewidth=2,
    label="Fitted Log-Normal PDF"
)
axes[0].set_title("PDF of Transaction Amounts")
axes[0].set_xlabel("Transaction Amount (₹)")
axes[0].set_ylabel("Probability Density")
axes[0].legend()
axes[0].grid(alpha=0.3)

# CDF plot: observed empirical CDF + fitted CDF
sorted_amounts = np.sort(amounts_clean)
empirical_cdf = np.arange(1, len(sorted_amounts) + 1) / len(sorted_amounts)

axes[1].step(
    sorted_amounts,
    empirical_cdf,
    where="post",
    label="Empirical CDF"
)
axes[1].plot(
    x, cdf_values,
    linewidth=2,
    label="Fitted Log-Normal CDF"
)
axes[1].set_title("CDF of Transaction Amounts")
axes[1].set_xlabel("Transaction Amount (₹)")
axes[1].set_ylabel("Cumulative Probability")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
threshold = 5000

model_cdf = lognorm.cdf(
    threshold,
    shape_ln,
    loc=loc_ln,
    scale=scale_ln
)

observed_cdf = (amounts_clean <= threshold).mean()

print(f"Model P(Amount <= ₹5000): {model_cdf:.4f}")
print(f"Model P(Amount > ₹5000):  {1 - model_cdf:.4f}")
print(f"Observed P(Amount <= ₹5000): {observed_cdf:.4f}")
print(f"Observed P(Amount > ₹5000):  {(amounts_clean > threshold).mean():.4f}")
```

##### Conclusion: PDF and CDF Analysis

The Probability Density Function (PDF) and Cumulative Distribution Function (CDF) were plotted for transaction amounts using a fitted Log-Normal distribution.

At the ₹5,000 threshold, the fitted model estimated a cumulative probability of 0.8616, meaning approximately 86.16% of transactions were expected to be ₹5,000 or less. The observed proportion was 88.64%.

The model estimated a 13.84% probability of a transaction exceeding ₹5,000, compared with the observed proportion of 11.36%. The difference of 2.48 percentage points indicates that the model slightly overestimated the exceedance probability at this threshold.

Overall, the PDF and CDF provide useful ways to understand transaction density and cumulative risk. However, these results describe the fit at one threshold and do not establish that the Log-Normal distribution fits the entire dataset perfectly.
